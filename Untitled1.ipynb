```python
"""
Early-Window Mass Prediction System for Retail Smart Cart Scale (0-50 kg)
Author: AI Collaborator / Engineering Team
Description: Trains LightGBM point and quantile regressors across multiple
early-window durations from transient load cell ringing curves. Implements
rigorous baselines, confidence cascade evaluation, stress testing, and
inference latency benchmarking.
"""

import os
import time
import numpy as np
import pandas as pd
import lightgbm as lgb
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.metrics import mean_absolute_error

# =========================================================================
# 1. CONFIGURATION BLOCK
# =========================================================================
CONFIG = {
    "data_dir": "/content/LOAD_CELL_DATASET",
    "main_csv": "load_cell_main.csv",  # Matches your file exactly
    "main_meta": "load_cell_main_meta.csv",
    "stress_csv": (
        "load_cell_heavy_preload.csv"
    ),  # Or load_cell_stress.csv if available
    "stress_meta": "load_cell_heavy_preload_meta.csv",
    "step_threshold": 15.0,
    "pre_samples": 5,
    "window_lengths": [4, 8, 12, 16, 24, 40],
    "window_labels": ["50ms", "100ms", "150ms", "200ms", "300ms", "500ms"],
    "train_ratio": 0.70,
    "val_ratio": 0.15,
    "test_ratio": 0.15,
    "random_seed": 42,
    "cascade_thresholds": [0.05, 0.10, 0.15, 0.20, 0.30, 0.40, 0.50, 0.60],
}

# =========================================================================
# 2. DATA LOADING & PREPROCESSING UTILITIES
# =========================================================================
def load_dataset(csv_path, meta_path):
    print(f"Loading {csv_path} and {meta_path}...")
    df = pd.read_csv(csv_path)
    meta = pd.read_csv(meta_path)
    y = df.iloc[:, 0].values            # True Mass in grams
    X_raw = df.iloc[:, 1:].values       # Raw time series readings
    return y, X_raw, meta

def preprocess_and_extract_features(X_raw, y_raw, meta_raw, window_n, config):
    """
    Preprocesses curves, detects steps, extracts windows and engineered features,
    and discards invalid rows.
    """
    valid_indices = []
    processed_features = []
    processed_y = []
    processed_meta = []
    onset_times = []

    dropped_no_step = 0
    dropped_out_of_bounds = 0

    pre_samples = config["pre_samples"]
    threshold = config["step_threshold"]

    for i in range(len(X_raw)):
        row = X_raw[i]

        # 1. Baseline subtraction (mean of first pre_samples)
        baseline = np.mean(row[:pre_samples])
        detrended = row - baseline

        # 2. Step detection from signal only
        exceeds = np.where(detrended > threshold)[0]
        if len(exceeds) == 0:
            dropped_no_step += 1
            continue
        s = exceeds[0]

        # 3. Check window bounds
        if s + window_n > len(row):
            dropped_out_of_bounds += 1
            continue

        # Extract window
        window = detrended[s : s + window_n]
        time_axis = np.arange(window_n) / 80.0  # 80 SPS

        # 4. Feature Engineering
        mean_val = np.mean(window)
        std_val  = np.std(window)
        max_val  = np.max(window)
        min_val  = np.min(window)

        # Slope via linear fit (polyfit degree 1)
        if window_n > 1:
            slope = np.polyfit(time_axis, window, 1)[0]
        else:
            slope = 0.0

        peak_idx = float(np.argmax(window))
        peak_val = float(np.max(window))
        last_4_mean = np.mean(window[-min(4, window_n):])

        # Combine raw window values and engineered features into a single feature vector
        features = np.concatenate([
            window,
            [mean_val, std_val, max_val, min_val, slope, peak_idx, peak_val, last_4_mean]
        ])

        valid_indices.append(i)
        processed_features.append(features)
        processed_y.append(y_raw[i])
        processed_meta.append(meta_raw.iloc[i].to_dict())
        onset_times.append(s / 80.0)

    print(f"  [Window {window_n} pts] Valid rows: {len(valid_indices)} | Dropped (no step): {dropped_no_step} | Dropped (out of bounds): {dropped_out_of_bounds}")

    return (
        np.array(processed_features),
        np.array(processed_y),
        pd.DataFrame(processed_meta),
        np.array(onset_times),
        valid_indices
    )

# =========================================================================
# 3. EVALUATION METRICS HELPERS
# =========================================================================
def compute_metrics(y_true, y_pred):
    y_pred = np.clip(y_pred, 1.0, 100000.0)  # Handle zero/negative predictions
    mae = mean_absolute_error(y_true, y_pred)
    rel_err = 100.0 * np.abs(y_pred - y_true) / y_true
    median_re = np.median(rel_err)
    p95_re = np.percentile(rel_err, 95)
    mape = np.mean(rel_err)
    return {
        "mae": mae,
        "median_re": median_re,
        "p95_re": p95_re,
        "mape": mape
    }

# =========================================================================
# 4. MAIN EXECUTION PIPELINE
# =========================================================================
def main():
    np.random.seed(CONFIG["random_seed"])
    data_dir = CONFIG["data_dir"]

    # Load main and stress datasets using configured paths
    y_main_raw, X_main_raw, meta_main_raw = load_dataset(
        os.path.join(data_dir, CONFIG["main_csv"]),
        os.path.join(data_dir, CONFIG["main_meta"])
    )
    y_stress_raw, X_stress_raw, meta_stress_raw = load_dataset(
        os.path.join(data_dir, CONFIG["stress_csv"]),
        os.path.join(data_dir, CONFIG["stress_meta"])
    )

    # Create consistent row split indices on the main dataset
    indices = np.arange(len(y_main_raw))
    train_val_idx, test_idx = train_test_split(
        indices, test_size=CONFIG["test_ratio"], random_state=CONFIG["random_seed"]
    )
    rel_val_size = CONFIG["val_ratio"] / (CONFIG["train_ratio"] + CONFIG["val_ratio"])
    train_idx, val_idx = train_test_split(
        train_val_idx, test_size=rel_val_size, random_state=CONFIG["random_seed"]
    )

    print(f"Dataset splits -> Train: {len(train_idx)}, Val: {len(val_idx)}, Test: {len(test_idx)}")

    results_summary = []
    models_dict = {}
    test_evals_per_window = {}
    stress_evals_per_window = {}

    # Loop over each window length
    for idx, w_len in enumerate(CONFIG["window_lengths"]):
        w_label = CONFIG["window_labels"].get(idx) if isinstance(CONFIG["window_labels"], dict) else CONFIG["window_labels"][idx]
        print(f"\n--- Processing Window: {w_label} ({w_len} samples) ---")

        # Process Main Dataset
        X_m, y_m, meta_m, onset_m, valid_m = preprocess_and_extract_features(
            X_main_raw, y_main_raw, meta_main_raw, w_len, CONFIG
        )
        # Process Stress Dataset
        X_s, y_s, meta_s, onset_s, valid_s = preprocess_and_extract_features(
            X_stress_raw, y_stress_raw, meta_stress_raw, w_len, CONFIG
        )

        # Filter split indices to only include valid rows
        valid_m_arr = np.array(valid_m)
        valid_s_arr = np.array(valid_s)

        tr_mask = np.isin(valid_m_arr, train_idx)
        va_mask = np.isin(valid_m_arr, val_idx)
        te_mask = np.isin(valid_m_arr, test_idx)

        X_tr, y_tr = X_m[tr_mask], y_m[tr_mask]
        X_va, y_va = X_m[va_mask], y_m[va_mask]
        X_te, y_te = X_m[te_mask], y_m[te_mask]
        meta_te = meta_m.iloc[te_mask].reset_index(drop=True)

        stress_mask = np.isin(valid_s_arr, np.arange(len(y_stress_raw)))
        X_str, y_str = X_s[stress_mask], y_s[stress_mask]
        meta_str = meta_s.iloc[stress_mask].reset_index(drop=True)

        # Log-transform target for ML models
        y_tr_log = np.log(y_tr)
        y_va_log = np.log(y_va)

        # ---------------------------------------------------------------------
        # BASELINE A: Window Mean Prediction
        # ---------------------------------------------------------------------
        pred_baseA_te = np.mean(X_te[:, :w_len], axis=1)
        pred_baseA_te = np.clip(pred_baseA_te, 20.0, 50000.0)
        metrics_baseA_te = compute_metrics(y_te, pred_baseA_te)

        pred_baseA_str = np.mean(X_str[:, :w_len], axis=1)
        pred_baseA_str = np.clip(pred_baseA_str, 20.0, 50000.0)
        metrics_baseA_str = compute_metrics(y_str, pred_baseA_str)

        # ---------------------------------------------------------------------
        # BASELINE B: Settled Reading (mean of last 0.3s / 24 samples of full record)
        # ---------------------------------------------------------------------
        def get_baseline_b(X_raw_subset, valid_subset):
            preds = []
            for original_idx in valid_subset:
                row = X_raw_subset[original_idx]
                baseline = np.mean(row[:CONFIG["pre_samples"]])
                settled = np.mean(row[-24:]) - baseline  # last 0.3s (24 samples @ 80SPS)
                preds.append(max(20.0, settled))
            return np.array(preds)

        pred_baseB_te = get_baseline_b(X_main_raw, valid_m_arr[te_mask])
        metrics_baseB_te = compute_metrics(y_te, pred_baseB_te)

        pred_baseB_str = get_baseline_b(X_stress_raw, valid_s_arr[stress_mask])
        metrics_baseB_str = compute_metrics(y_str, pred_baseB_str)

        # ---------------------------------------------------------------------
        # LIGHTGBM POINT REGRESSOR (Predicting log(mass))
        # ---------------------------------------------------------------------
        train_data = lgb.Dataset(X_tr, label=y_tr_log)
        val_data = lgb.Dataset(X_va, label=y_va_log, reference=train_data)

        params = {
            'objective': 'regression',
            'metric': 'rmse',
            'boosting_type': 'gbdt',
            'num_leaves': 31,
            'learning_rate': 0.05,
            'feature_fraction': 0.8,
            'verbose': -1,
            'random_state': CONFIG["random_seed"]
        }

        model_point = lgb.train(
            params,
            train_data,
            num_boost_round=1000,
            valid_sets=[val_data],
            callbacks=[lgb.early_stopping(50, verbose=False)]
        )

        pred_lgb_log_te = model_point.predict(X_te)
        pred_lgb_te = np.exp(pred_lgb_log_te)
        metrics_lgb_te = compute_metrics(y_te, pred_lgb_te)

        pred_lgb_log_str = model_point.predict(X_str)
        pred_lgb_str = np.exp(pred_lgb_log_str)
        metrics_lgb_str = compute_metrics(y_str, pred_lgb_str)

        # ---------------------------------------------------------------------
        # LIGHTGBM QUANTILE REGRESSORS (alpha = 0.1 and 0.9)
        # ---------------------------------------------------------------------
        def train_quantile(alpha):
            q_params = {
                'objective': 'quantile',
                'alpha': alpha,
                'metric': 'quantile',
                'boosting_type': 'gbdt',
                'num_leaves': 31,
                'learning_rate': 0.05,
                'feature_fraction': 0.8,
                'verbose': -1,
                'random_state': CONFIG["random_seed"]
            }
            return lgb.train(
                q_params,
                train_data,
                num_boost_round=1000,
                valid_sets=[val_data],
                callbacks=[lgb.early_stopping(50, verbose=False)]
            )

        model_q10 = train_quantile(0.1)
        model_q90 = train_quantile(0.9)

        q10_te = np.exp(model_q10.predict(X_te))
        q90_te = np.exp(model_q90.predict(X_te))

        q10_str = np.exp(model_q10.predict(X_str))
        q90_str = np.exp(model_q90.predict(X_str))

        # Save models
        models_dict[w_len] = {
            "point": model_point,
            "q10": model_q10,
            "q90": model_q90
        }

        # Store evaluation structures for cascade & plotting
        test_evals_per_window[w_len] = {
            "y_true": y_te,
            "pred": pred_lgb_te,
            "q10": q10_te,
            "q90": q90_te,
            "onset": onset_m[te_mask],
            "meta": meta_te,
            "X": X_te
        }

        stress_evals_per_window[w_len] = {
            "y_true": y_str,
            "pred": pred_lgb_str,
            "q10": q10_str,
            "q90": q90_str,
            "onset": onset_s[stress_mask],
            "meta": meta_str,
            "X": X_str
        }

        results_summary.append({
            "Window": w_label,
            "LGB_Test_MAE_g": metrics_lgb_te["mae"],
            "LGB_Test_MedRE_%": metrics_lgb_te["median_re"],
            "LGB_Test_P95RE_%": metrics_lgb_te["p95_re"],
            "BaseA_MedRE_%": metrics_baseA_te["median_re"],
            "BaseB_MedRE_%": metrics_baseB_te["median_re"],
            "Stress_MedRE_%": metrics_lgb_str["median_re"],
            "Stress_P95RE_%": metrics_lgb_str["p95_re"]
        })

        print(f"  -> LGB Test MAE: {metrics_lgb_te['mae']:.2f}g | Median RE: {metrics_lgb_te['median_re']:.2f}% | P95 RE: {metrics_lgb_te['p95_re']:.2f}%")

    results_df = pd.DataFrame(results_summary)
    print("\n================ RESULTS TABLE ================ ")
    print(results_df.to_string(index=False))
    results_df.to_csv("results.csv", index=False)
    print("Saved results table to results.csv")

    # =========================================================================
    # 5. CONFIDENCE CASCADE EVALUATION
    # =========================================================================
    print("\n--- Running Confidence Cascade Threshold Sweep ---")
    cascade_results = []
    window_sorted = CONFIG["window_lengths"]
    n_test_samples = len(test_evals_per_window[window_sorted[0]]["y_true"])

    for thresh in CONFIG["cascade_thresholds"]:
        accepted_preds = np.zeros(n_test_samples)
        decision_times = np.zeros(n_test_samples)
        acceptance_counts = np.zeros(len(window_sorted))

        for i in range(n_test_samples):
            accepted = False
            for w_idx, w_len in enumerate(window_sorted):
                eval_data = test_evals_per_window[w_len]
                pred = eval_data["pred"][i]
                q10  = eval_data["q10"][i]
                q90  = eval_data["q90"][i]
                onset = eval_data["onset"][i]
                w_dur = w_len / 80.0

                # Interval width ratio
                interval_width = (q90 - q10) / max(1.0, pred)

                if interval_width <= thresh or w_idx == len(window_sorted) - 1:
                    accepted_preds[i] = pred
                    decision_times[i] = onset + w_dur
                    acceptance_counts[w_idx] += 1
                    accepted = True
                    break

        y_true_test = test_evals_per_window[window_sorted[0]]["y_true"]
        cascade_metrics = compute_metrics(y_true_test, accepted_preds)
        mean_time_ms = np.mean(decision_times) * 1000.0
        acceptance_pcts = (acceptance_counts / n_test_samples) * 100.0

        cascade_results.append({
            "Threshold": thresh,
            "Mean_Decision_Time_ms": mean_time_ms,
            "Test_MAE_g": cascade_metrics["mae"],
            "Test_MedRE_%": cascade_metrics["median_re"],
            "Test_P95RE_%": cascade_metrics["p95_re"],
            "Acceptance_50ms_%": acceptance_pcts[0],
            "Acceptance_500ms_%": acceptance_pcts[-1]
        })

    cascade_df = pd.DataFrame(cascade_results)
    print(cascade_df.to_string(index=False))

    # =========================================================================
    # 6. GENERATE & SAVE PLOTS (PNG)
    # =========================================================================
    print("\n--- Generating Plots ---")
    os.makedirs(".", exist_ok=True)

    # Plot 1: Relative Error vs Window Length
    plt.figure(figsize=(9, 5))
    w_labels = CONFIG["window_labels"]
    lgb_meds, lgb_p95s, stress_meds = [], [], []
    for w in window_sorted:
        te_m = compute_metrics(test_evals_per_window[w]["y_true"], test_evals_per_window[w]["pred"])
        str_m = compute_metrics(stress_evals_per_window[w]["y_true"], stress_evals_per_window[w]["pred"])
        lgb_meds.append(te_m["median_re"])
        lgb_p95s.append(te_m["p95_re"])
        stress_meds.append(str_m["median_re"])

    plt.plot(w_labels, lgb_meds, marker='o', linewidth=2, label="LGB Test Median RE (%)")
    plt.plot(w_labels, lgb_p95s, marker='s', linewidth=2, linestyle='--', label="LGB Test P95 RE (%)")
    plt.plot(w_labels, stress_meds, marker='^', linewidth=2, linestyle=':', label="Stress Set Median RE (%)")
    plt.xlabel("Window Length")
    plt.ylabel("Relative Error (%)")
    plt.title("Relative Error vs. Window Length")
    plt.grid(True, alpha=0.3)
    plt.legend()
    plt.tight_layout()
    plt.savefig("plot_1_rel_error_vs_window.png", dpi=300)
    plt.close()

    # Plot 2: MAE by Mass Band (using 300ms / 24 pts)
    best_w = 24 if 24 in test_evals_per_window else window_sorted[-2]
    eval_b = test_evals_per_window[best_w]
    y_t = eval_b["y_true"]
    y_p = eval_b["pred"]
    bands = [(20, 100, "20-100g"), (100, 500, "100-500g"), (500, 2000, "0.5-2kg"), (2000, 5000, "2-5kg"), (5000, 10000, "5-10kg")]
    band_labels, band_maes = [], []
    for low, high, lbl in bands:
        mask = (y_t >= low) & (y_t < high)
        if np.sum(mask) > 0:
            band_maes.append(mean_absolute_error(y_t[mask], y_p[mask]))
            band_labels.append(lbl)

    plt.figure(figsize=(8, 5))
    plt.bar(band_labels, band_maes, color='skyblue', edgecolor='navy')
    plt.xlabel("Mass Band")
    plt.ylabel("MAE (grams)")
    plt.title(f"Test MAE by Mass Band (Window: {best_w} samples)")
    plt.grid(axis='y', alpha=0.3)
    plt.tight_layout()
    plt.savefig("plot_2_mae_by_mass_band.png", dpi=300)
    plt.close()

    # Plot 3: Relative Error vs Noise & Vibration
    meta_df = eval_b["meta"]
    if "NoiseSigma_g" in meta_df.columns and "VibAmp_g" in meta_df.columns:
        rel_errs = 100.0 * np.abs(y_p - y_t) / y_t
        meta_df["RelErr"] = rel_errs

        fig, axes = plt.subplots(1, 2, figsize=(12, 4))
        meta_df["NoiseBin"] = pd.qcut(meta_df["NoiseSigma_g"], q=4, duplicates='drop')
        meta_df.groupby("NoiseBin")["RelErr"].median().plot(kind='bar', ax=axes[0], color='orange', edgecolor='darkred')
        axes[0].set_title("Median RE (%) vs. Noise Sigma")
        axes[0].set_ylabel("Median RE (%)")
        axes[0].grid(axis='y', alpha=0.3)

        meta_df["VibBin"] = pd.qcut(meta_df["VibAmp_g"], q=4, duplicates='drop')
        meta_df.groupby("VibBin")["RelErr"].median().plot(kind='bar', ax=axes[1], color='teal', edgecolor='darkblue')
        axes[1].set_title("Median RE (%) vs. Vibration Amplitude")
        axes[1].set_ylabel("Median RE (%)")
        axes[1].grid(axis='y', alpha=0.3)

        plt.tight_layout()
        plt.savefig("plot_3_error_vs_noise_vib.png", dpi=300)
        plt.close()

    # Plot 4: Cascade Trade-off (Mean Decision Time vs P95 RE)
    plt.figure(figsize=(8, 5))
    plt.plot(cascade_df["Mean_Decision_Time_ms"], cascade_df["Test_P95RE_%"], marker='o', color='purple', linewidth=2)
    for i, row in cascade_df.iterrows():
        plt.text(row["Mean_Decision_Time_ms"], row["Test_P95RE_%"] + 0.2, f"T={row['Threshold']}", fontsize=9)
    plt.xlabel("Mean Decision Time (ms)")
    plt.ylabel("Test P95 Relative Error (%)")
    plt.title("Confidence Cascade: Speed vs. Accuracy Trade-off")
    plt.grid(True, alpha=0.3)
    plt.tight_layout()
    plt.savefig("plot_4_cascade_tradeoff.png", dpi=300)
    plt.close()

    # Plot 5: Predicted vs True Mass Scatter (Log Axes)
    plt.figure(figsize=(6, 6))
    plt.scatter(y_t, y_p, alpha=0.3, color='royalblue', s=15)
    plt.plot([20, 10000], [20, 10000], color='red', linestyle='--', label='Ideal')
    plt.xscale('log')
    plt.yscale('log')
    plt.xlabel("True Mass (g)")
    plt.ylabel("Predicted Mass (g)")
    plt.title(f"Predicted vs True Mass (Window: {best_w} samples)")
    plt.legend()
    plt.grid(True, which="both", alpha=0.2)
    plt.tight_layout()
    plt.savefig("plot_5_pred_vs_true_scatter.png", dpi=300)
    plt.close()
    print("All 5 plots saved successfully as PNGs.")

    # =========================================================================
    # 7. DEPLOYMENT: MODEL SAVING & LATENCY BENCHMARKING
    # =========================================================================
    print("\n--- Model Saving & Inference Latency Benchmarking ---")
    sample_features = test_evals_per_window[best_w]["X"][0].reshape(1, -1)

    # Save models for the best window
    best_models = models_dict[best_w]
    best_models["point"].save_model("load_cell_point_model.txt")
    best_models["q10"].save_model("load_cell_q10_model.txt")
    best_models["q90"].save_model("load_cell_q90_model.txt")
    print("Saved models: load_cell_point_model.txt, load_cell_q10_model.txt, load_cell_q90_model.txt")

    # Benchmark single-row inference latency (median of 1000 calls)
    def benchmark_latency(model, n_calls=1000):
        times = []
        for _ in range(n_calls):
            t0 = time.perf_counter()
            _ = model.predict(sample_features)
            t1 = time.perf_counter()
            times.append((t1 - t0) * 1000.0)  # ms
        return np.median(times)

    lat_point = benchmark_latency(best_models["point"])
    lat_q10   = benchmark_latency(best_models["q10"])
    lat_q90   = benchmark_latency(best_models["q90"])

    print(f"Single-row inference latency (Median of 1000 calls @ Window {best_w}):")
    print(f"  - Point Regressor: {lat_point:.4f} ms")
    print(f"  - Quantile 10%:    {lat_q10:.4f} ms")
    print(f"  - Quantile 90%:    {lat_q90:.4f} ms")
    print("\nPipeline execution complete successfully!")

if __name__ == "__main__":
    main()
```
